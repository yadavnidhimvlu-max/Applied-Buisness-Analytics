Part A

```python
import pandas as pd
from scipy.stats import ttest_1samp
df = pd.read_csv("supermarket_sales.csv")
df.columns = df.columns.str.strip()
```

```python
print("First 5 Records:")
print(df.head())
revenue = df["Total"]
```

```python

print("\nMissing Values:")
print(revenue.isnull().sum())
revenue = revenue.dropna()


print("\nMean Revenue:")
print(revenue.mean())
```

```python
expected_mean = 1500
# H0: μ = 1500
# H1: μ ≠ 1500
alpha = 0.05
t_stat, p_value = ttest_1samp(revenue, expected_mean)
```

```python
print("\nT-statistic:", t_stat)
print("P-value:", p_value)
print("Nidhi Yadav T053")
if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Null Hypothesis is not rejected")
if p_value < alpha:
    print("The average revenue is significantly different from ₹1500.")
else:
    print("There is not enough evidence to conclude that average revenue differs from ₹1500.")
```

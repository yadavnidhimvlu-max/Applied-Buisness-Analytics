```python
import pandas as pd

df = pd.read_csv("supermarket_sales.csv")
print(df.head(10))
print("Nidhi  Yadav  T053")
```

```python
print("Number of Records:", len(df))
print("Number of Columns:", len(df.columns))
print("Column Names:", df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
print(" Nidhi Yadav T053")
```

```python
print(df["Product line"].unique())
```

```python
total_revenue = df.groupby("Product line")["Total"].sum()
print(total_revenue)
```

```python
avg_revenue = df.groupby("Product line")["Total"].mean()
print(avg_revenue)
```

```python
total_quantity = df.groupby("Product line")["Quantity"].sum()
print(total_quantity)
```

```python
median_revenue = df["Total"].median()
print("Median Revenue:", median_revenue)
```

```python
df["revenue_level"] = df["Total"].apply(
    lambda x: "High Revenue" if x >= median_revenue else "Low Revenue"
)
```

```python
print(df[["Product line", "Total", "revenue_level"]].head(10))
```

```python
count = pd.crosstab(df["Product line"], df["revenue_level"])
print(count)
```

```python
probability = pd.crosstab(
    df["Product line"],
    df["revenue_level"],
    normalize="index"
)

print(probability)
print("Nidhi Yadav T053")
```

```python
print(probability.sum(axis=1))
```

```python
payoff = df.groupby(
    ["Product line", "revenue_level"]
)["Total"].mean()

print(payoff)
```

```python
decision_df = payoff.reset_index()
decision_df["Probability"] = decision_df.apply(
    lambda row: probability.loc[
        row["Product line"], row["revenue_level"]
    ],
    axis=1
)

decision_df = decision_df.rename(columns={"Total": "Payoff"})

print(decision_df)
```

```python
print(decision_df[
    ["Product line", "revenue_level", "Probability", "Payoff"]
])
```

```python
ev = decision_df.copy()

ev["Probability_Payoff"] = (
    ev["Probability"] * ev["Payoff"]
)

expected_value = ev.groupby(
    "Product line"
)["Probability_Payoff"].sum()

print(expected_value)
```

```python
print(expected_value.sort_values(ascending=False))
print("Nidhi Yadav T053")
```

```python
highest = expected_value.idxmax()
print("Highest Expected Value:", highest)
```

```python
lowest = expected_value.idxmin()
print("Lowest Expected Value:", lowest)
```

```python
ranking = expected_value.sort_values(ascending=False)
print(ranking)
```

```python
highest_probability_product = probability["High Revenue"].idxmax()
highest_ev_product = expected_value.idxmax()

print("Highest High Revenue Probability:",
      highest_probability_product)

print("Highest Expected Value:",
      highest_ev_product)

if highest_probability_product == highest_ev_product:
    print("Both are the same.")
else:
    print("Both are different.")
```

```python
high_probability = 0.60
new_high_payoff = 600
low_probability = 0.40
low_payoff = 200

new_ev = (
    high_probability * new_high_payoff
    + low_probability * low_payoff
)

print("New Expected Value:", new_ev)
print(" Nidhi  Yadav  T053")
```

# Practical 1B

```python
#Import Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load dataset
df = pd.read_csv("Retail_sales - Retail_sales.csv")
```

Analysis 1 - Product category analysis

```python
print(df["Product Category"].value_counts())
```

```python
df["Product Category"].value_counts().plot(kind="bar")
plt.title("Product Category Analysis")
plt.xlabel("Product Category")
plt.ylabel("Number of Sales")
plt.show()
```

Analysis 2 - Store Location Analysis

```python
print(df["Store Location"].value_counts())
```

```python
df["Store Location"].value_counts().plot(kind="bar")
plt.title("Store Location Analysis")
plt.xlabel("Store Location")
plt.ylabel("Number of Sales")
plt.show()
```

Analysis 3 - Day of the Week Analysis

```python
print(df["Day of the Week"].value_counts())
```

```python
df["Day of the Week"].value_counts().plot(kind="bar")
plt.title("Sales by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Number of Sales")
plt.show()
```

Analysis 4 - Holiday Effect Analysis

```python
print(df["Holiday Effect"].value_counts())
```

```python
df["Holiday Effect"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Holiday Effect")
plt.ylabel("")
plt.show()
```

Analysis 5 - Sales Revenue Analysis

```python
plt.hist(df["Sales Revenue (USD)"], bins=10)
plt.title("Sales Revenue Distribution")
plt.xlabel("Sales Revenue")
plt.ylabel("Frequency")
plt.show()
```

Analysis 6 - Units Sold Analysis

```python
plt.hist(df["Units Sold"], bins=10)
plt.title("Units Sold Distribution")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()
```

Analysis 7 - Marketing Spend vs Sales Revenue

```python
plt.scatter(df["Marketing Spend (USD)"], df["Sales Revenue (USD)"])
plt.title("Marketing Spend vs Sales Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales Revenue")
plt.show()
```

Analysis 8 - Correlation Analysis

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Reds")
plt.title("Correlation Heatmap")
plt.show()
```
